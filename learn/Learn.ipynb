```python
#最短的仿真程序
#导入物理引擎
import mujoco
#xml="""..."""一个字符串，描述场景，装着机器人的硬件
#<worldbody>场景的根节点，机器人的硬件设施安装再里面
#<geom type="plane" size="5 5 .1"/> 一个平面，长宽各5米，形成地面，geom是几何体
#<body pos="0 0 2">一个刚体，放在世界坐标(0,0,2),也就是离地面2米高
#<freejoint/>自由关节,让这个刚体可以在空间中自由平移和旋转，没有关节的body是焊死在世界上，不会动
#<geom type="box" size=".2 .2 .2"/>这个刚体是一个边长0.4米的立方体，size是边长
xml = """
<mujoco>
    <worldbody>
        <geom type="plane" size="5 5 .1"/>
        <body pos="0 0 2">
           <freejoint/>
           <geom type="box" size=".2 .2 .2"/>
        </body>
    </worldbody>
</mujoco>
"""

#把xml解析成模型，model是只读的静态描述：有几个关节，多重，有没有电机，建好之后不再改变
model = mujoco.MjModel.from_xml_string(xml)
#创建运行时的数据，data状态发生改变，每个关节当前的角度、速度，初始全部都是0
data = mujoco.MjData(model)

#推进1000步仿真，默认每步0.002秒，1000步就是2秒的仿真时间
for i in range(1000):
    mujoco.mj_step(model, data)

print(data.qpos)
```
